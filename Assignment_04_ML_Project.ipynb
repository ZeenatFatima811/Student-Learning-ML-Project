### Dataset Inspection

The student_learning.csv dataset contains 280 student records and 7 columns.

The dataset includes numerical and categorical features such as hours_studied, attendance, practice_tests, sleep_hours and study_method.

The target variable is passed:
- 0 represents Fail.
- 1 represents Pass.

The target distribution contains 157 students who failed and 123 students who passed.

There are no missing values in the dataset.

The final_score column will be excluded from model training because it reveals information about the final result and can cause data leakage.

```python
# Select input features (X)
X = df.drop(columns=['passed', 'final_score'])

# Select target variable (y)
y = df['passed']

# Display selected features
print("Selected Features:")
print(X.head())

print("\nFeature Names:")
print(X.columns.tolist())

print("\nTarget Variable:")
print(y.name)
```

```python
assert 'final_score' not in X.columns
assert 'passed' not in X.columns

print("Feature selection successful!")
print("No data leakage columns included.")
```

### Feature Selection and Data Leakage Prevention

The target variable is passed, which represents whether a student passed or failed.

The input features selected for training are:
- hours_studied
- attendance
- practice_tests
- sleep_hours
- study_method

The final_score column has been excluded because it contains information directly related to the final outcome. Including it could cause data leakage and result in unrealistic model performance.

Therefore, only information available before the final result is used for prediction.

```python
from sklearn.model_selection import train_test_split

print("Train-test split library imported successfully!")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)

print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)
```

```python
print("Original Target Distribution:")
print(y.value_counts(normalize=True) * 100)

print("\nTraining Target Distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting Target Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

### Train-Test Split

The dataset was divided into 80% training data and 20% testing data.

A random_state of 42 was used to ensure reproducibility.

Stratified sampling was applied using stratify=y to maintain a similar proportion of passed and failed students in both sets.

The training set contains 224 records, while the testing set contains 56 records.

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix

print("Baseline model libraries imported successfully!")
```

```python
baseline_model = DummyClassifier(strategy='most_frequent')

baseline_model.fit(X_train, y_train)

print("Baseline model trained successfully!")
```

```python
baseline_predictions = baseline_model.predict(X_test)

print("Actual Values:")
print(y_test.values[:10])

print("\nPredicted Values:")
print(baseline_predictions[:10])
```

```python
baseline_accuracy = accuracy_score(y_test, baseline_predictions)
baseline_f1 = f1_score(y_test, baseline_predictions, zero_division=0)
baseline_cm = confusion_matrix(y_test, baseline_predictions)

print("Baseline Accuracy:", baseline_accuracy)
print("Baseline F1 Score:", baseline_f1)

print("\nBaseline Confusion Matrix:")
print(baseline_cm)
```

### Baseline Model

A DummyClassifier with the most_frequent strategy was trained as the baseline model.

This model always predicts the majority class from the training dataset.

The baseline model provides a simple performance benchmark. More advanced machine learning models should be evaluated against this baseline using the same test dataset.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

print("Machine Learning libraries imported successfully!")
```

```python
numeric_features = [
    'hours_studied',
    'attendance',
    'practice_tests',
    'sleep_hours'
]

categorical_features = ['study_method']

print("Numerical Features:", numeric_features)
print("Categorical Features:", categorical_features)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

print("Preprocessing setup completed!")
```

```python
logistic_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', LogisticRegression(max_iter=1000))
    ]
)

logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
random_forest_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', RandomForestClassifier(
            n_estimators=200,
            random_state=42
        ))
    ]
)

random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

### Model Training and Preprocessing

Two machine learning models were trained:

1. Logistic Regression
2. Random Forest Classifier

A ColumnTransformer was used for preprocessing.

StandardScaler was applied to numerical features, while OneHotEncoder was applied to the categorical study_method feature.

Both preprocessing and classification steps were combined using Scikit-learn Pipelines.

The models were trained using the same training dataset.

```python
# Logistic Regression predictions
logistic_predictions = logistic_model.predict(X_test)

# Random Forest predictions
rf_predictions = random_forest_model.predict(X_test)

print("Both models generated predictions successfully!")
```

```python
from sklearn.metrics import accuracy_score, f1_score

# Logistic Regression evaluation
logistic_accuracy = accuracy_score(y_test, logistic_predictions)
logistic_f1 = f1_score(y_test, logistic_predictions)

# Random Forest evaluation
rf_accuracy = accuracy_score(y_test, rf_predictions)
rf_f1 = f1_score(y_test, rf_predictions)

print("LOGISTIC REGRESSION")
print("Accuracy:", logistic_accuracy)
print("F1 Score:", logistic_f1)

print("\nRANDOM FOREST")
print("Accuracy:", rf_accuracy)
print("F1 Score:", rf_f1)
```

```python
from sklearn.metrics import confusion_matrix

print("Logistic Regression Confusion Matrix:")
print(confusion_matrix(y_test, logistic_predictions))

print("\nRandom Forest Confusion Matrix:")
print(confusion_matrix(y_test, rf_predictions))
```

```python
from sklearn.metrics import classification_report

print("LOGISTIC REGRESSION REPORT")
print(classification_report(y_test, logistic_predictions))

print("\nRANDOM FOREST REPORT")
print(classification_report(y_test, rf_predictions))
```

```python
results = pd.DataFrame({
    'Model': ['Dummy Classifier', 'Logistic Regression', 'Random Forest'],
    'Accuracy': [
        baseline_accuracy,
        logistic_accuracy,
        rf_accuracy
    ],
    'F1 Score': [
        baseline_f1,
        logistic_f1,
        rf_f1
    ]
})

print(results)
```

```python
if logistic_f1 >= rf_f1:
    best_model = logistic_model
    best_model_name = "Logistic Regression"
else:
    best_model = random_forest_model
    best_model_name = "Random Forest"

print("Selected Best Model:", best_model_name)
```

### Model Evaluation and Comparison

The Dummy Classifier, Logistic Regression, and Random Forest models were evaluated using the same held-out test dataset.

Accuracy and F1 Score were used to compare model performance. Confusion matrices and classification reports were also generated.

The best candidate model was selected based on its F1 Score. Accuracy was considered as an additional performance metric.

The confusion matrices help identify false positive and false negative predictions.

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import joblib

# Load trained model
model = joblib.load('student_model.joblib')

# Page configuration
st.set_page_config(
    page_title="Student Pass Prediction",
    page_icon="🎓"
)

st.title("🎓 Student Learning Prediction App")

st.write(
    "Enter student learning details to predict "
    "the demonstration Pass/Fail result."
)

# User inputs
hours_studied = st.number_input(
    "Hours Studied",
    min_value=0.0,
    max_value=24.0,
    value=5.0
)

attendance = st.number_input(
    "Attendance (%)",
    min_value=0,
    max_value=100,
    value=75
)

practice_tests = st.number_input(
    "Practice Tests Completed",
    min_value=0,
    max_value=20,
    value=3
)

sleep_hours = st.number_input(
    "Sleep Hours",
    min_value=0.0,
    max_value=24.0,
    value=7.0
)

study_method = st.selectbox(
    "Study Method",
    ["self", "video", "group"]
)

# Prediction button
if st.button("Predict Result"):

    student_data = pd.DataFrame({
        'hours_studied': [hours_studied],
        'attendance': [attendance],
        'practice_tests': [practice_tests],
        'sleep_hours': [sleep_hours],
        'study_method': [study_method]
    })

    prediction = model.predict(student_data)[0]

    st.subheader("Prediction Result")

    if prediction == 1:
        st.success("Predicted Result: Pass")
    else:
        st.warning("Predicted Result: Fail")

    st.caption(
        "Demonstration only. This synthetic-data model "
        "must not be used to judge or penalize real students."
    )
```
